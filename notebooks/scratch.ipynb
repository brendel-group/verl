```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Specify the model name
model_name = "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"

# Load the tokenizer
print(f"Loading tokenizer for {model_name}...")
tokenizer = AutoTokenizer.from_pretrained(model_name)
print("Tokenizer loaded.")

# Load the model
print(f"Loading model {model_name}...")
# You might need to specify torch_dtype=torch.bfloat16 if running on GPUs that support it
# or device_map="auto" for automatic device placement
model = AutoModelForCausalLM.from_pretrained(model_name)
print("Model loaded.")

# You can now use the model and tokenizer
# For example, generate some text
prompt = "What is the capital of France?"
inputs = tokenizer(prompt, return_tensors="pt")

# Generate text
# Add device placement if you loaded the model to a specific device e.g. inputs.to(model.device)
outputs = model.generate(**inputs, max_new_tokens=50)
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("\nGenerated text:")
print(generated_text)
```
