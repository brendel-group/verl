```python
import pandas as pd
loc = '/fast/pmayilvahanan/datasets/deepscaler/test.parquet'
df = pd.read_parquet(loc)
```

```python
df_dsr = pd.read_parquet('/fast/pmayilvahanan/datasets/rl_training_one_example/dsr_sub.parquet')
df_dsr_no_instruction = pd.read_parquet('/fast/pmayilvahanan/datasets/rl_training_one_example/dsr_sub_no_instruction.parquet')

print(df_dsr.iloc[100]['prompt'])
print(df_dsr_no_instruction.iloc[100]['prompt'])
```

```python
# Create a new dataframe with modified prompts
df_dsr_modified = df_dsr.copy()

# Function to remove the instruction string from content
def remove_instruction(prompt_list):
    content = prompt_list[0]['content']
    # Remove the instruction string
    modified_content = content.replace(" Let's think step by step and output the final answer within \\boxed{}.", "")
    # Create new prompt with modified content
    return [{'content': modified_content, 'role': prompt_list[0].get('role', 'user')}]

# Apply the function to each row
df_dsr_modified['prompt'] = df_dsr['prompt'].apply(remove_instruction)

# Display the first modified prompt
df_dsr_modified.iloc[100]['prompt']
```

```python
df_dsr_modified.to_parquet('/fast/pmayilvahanan/datasets/rl_training_one_example/dsr_sub_no_instruction.parquet')
```

```python

# Create a mapping from df content to full_solution
content_to_solution = {}
for _, row in df.iterrows():
    prompt_content = row['prompt'][0]['content']
    if prompt_content and 'extra_info' in row and 'full_solution' in row['extra_info']:
        content_to_solution[prompt_content] = row['extra_info']['full_solution']

```

```python
# Update df_dsr extra_info with answers from df where content matches
for idx, row in df_dsr.iterrows():
    prompt_content = row['prompt'][0]['content']
    if prompt_content in content_to_solution:
        # Initialize extra_info as dict if it's not already
        if not isinstance(row['extra_info'], dict):
            df_dsr.at[idx, 'extra_info'] = {}
        
        # Create a copy to avoid modifying the original dict
        extra_info_dict = df_dsr.at[idx, 'extra_info'].copy() if isinstance(df_dsr.at[idx, 'extra_info'], dict) else {}
        extra_info_dict['answer'] = content_to_solution[prompt_content]
        df_dsr.at[idx, 'extra_info'] = extra_info_dict
```

```python
indices_with_answer = []
for idx, row in df_dsr.iterrows():
    if row['extra_info']['answer'] != '':
        print(row['extra_info']['answer'])
        indices_with_answer.append(idx)
        print(idx)
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Specify the model name
model_name = "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"

# Load the tokenizer
print(f"Loading tokenizer for {model_name}...")
tokenizer = AutoTokenizer.from_pretrained(model_name)
print("Tokenizer loaded.")

# Load the model
print(f"Loading model {model_name}...")
# You might need to specify torch_dtype=torch.bfloat16 if running on GPUs that support it
# or device_map="auto" for automatic device placement
model = AutoModelForCausalLM.from_pretrained(model_name)
print("Model loaded.")

# You can now use the model and tokenizer
# For example, generate some text
prompt = "What is the capital of France?"
inputs = tokenizer(prompt, return_tensors="pt")

# Generate text
# Add device placement if you loaded the model to a specific device e.g. inputs.to(model.device)
outputs = model.generate(**inputs, max_new_tokens=50)
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("\nGenerated text:")
print(generated_text)
```
