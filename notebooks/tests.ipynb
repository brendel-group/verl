Does math500 have overlap with math train?

```python
df_500 = pd.read_parquet('/fast/pmayilvahanan/datasets/openai_math/test.parquet')
df_math = pd.read_parquet('/fast/pmayilvahanan/datasets/openai_math/train.parquet')
```

```python
df_500
```

```python
# Extract the 'content' from the 'prompt' column for each DataFrame
prompts_math = [element[0]['content'] for element in df_math['prompt']]
prompts_500 = [element[0]['content'] for element in df_500['prompt']]
# Drop any None values that might have resulted from unexpected formats

# Convert the prompt Series to sets
set_math = set(prompts_math)
set_500 = set(prompts_500)

# Find the intersection
overlap = set_math.intersection(set_500)

# Print the number of overlapping prompts
print(f"Number of overlapping prompts: {len(overlap)}")
```

```python
# Extract the 'content' from the 'prompt' column for each DataFrame
solutions_math = [element['solution'] for element in df_math['extra_info']]
solutions_500 = [element['solution'] for element in df_500['extra_info']]
# Drop any None values that might have resulted from unexpectextra_infoed formats

# Convert the prompt Series to sets
set_math = set(solutions_math)
set_500 = set(solutions_500)

# Find the intersection
overlap = set_math.intersection(set_500)

# Print the number of overlapping prompts
print(f"Number of overlapping solutions: {len(overlap)}")
```
